# Ejercicio 1 - Importación de datos con Pandas

Combinar datos de ventas (CSV), clientes (JSON) e inventario (Excel) para un análisis de comercio electrónico.

```python
import pandas as pd

# 1. Cargar los datos de ventas desde el archivo CSV
ventas = pd.read_csv('ventas.csv')
print("Vista previa de VENTAS:")
ventas.head()
```

```python
# 2. Cargar los datos de clientes desde el archivo JSON
clientes = pd.read_json('clientes.json')
print("Cantidad total de clientes:", len(clientes))
clientes.head()
```

```python
# 3. Cargar los datos de inventario desde la hoja de cálculo Excel
inventario = pd.read_excel('inventario.xlsx')
promedio_stock = inventario['stock'].mean()
print("Promedio de stock disponible:", round(promedio_stock, 2))
inventario.head()
```

### Combinar los datos en una sola estructura

Unimos `ventas` con `clientes` (por `id_cliente`) para tener toda la información en un solo DataFrame.

```python
# Combinar ventas + clientes en una sola tabla
ventas_completas = ventas.merge(clientes, on='id_cliente', how='left')
ventas_completas.head()
```

```python
# Resumen general
print("Cantidad de ventas registradas:", len(ventas))
print("Cantidad de clientes:", len(clientes))
print("Cantidad de productos en inventario:", len(inventario))
print("Promedio de stock:", round(promedio_stock, 2))
```
