# Ejercicio 2 - Visualización con Matplotlib


```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Cargar el conjunto de datos
df = pd.read_csv('Automobile.csv')
df.head()
```

```python
df.info()
df.describe()
```

### 2. Gráfico de dispersión: mpg vs weight

Exploramos si existe relación entre el peso del auto y su consumo de combustible.

```python
plt.figure(figsize=(8,5))
plt.scatter(df['weight'], df['mpg'], alpha=0.6, color='teal')
plt.xlabel('Peso (weight)')
plt.ylabel('Consumo (mpg)')
plt.title('Relación entre peso y consumo de combustible')
plt.grid(True)
plt.show()
```

### 3. Histograma del consumo de combustible (mpg)

```python
plt.figure(figsize=(8,5))
plt.hist(df['mpg'], bins=15, color='steelblue', edgecolor='black')
plt.xlabel('Millas por galón (mpg)')
plt.ylabel('Frecuencia')
plt.title('Distribución del consumo de combustible')
plt.show()
```

### 4. Boxplot de cylinders y origin

Permite ver la dispersión de valores y detectar posibles outliers.

```python
plt.figure(figsize=(6,5))
plt.boxplot([df['cylinders'], df['origin']], tick_labels=['cylinders', 'origin'])
plt.title('Boxplot de cylinders y origin')
plt.show()
```

### 5. Gráfico de barras: autos por año de modelo y origen

origin: 1 = EE.UU., 2 = Europa, 3 = Asia

```python
conteo = df.groupby(['model_year', 'origin']).size().unstack(fill_value=0)

conteo.plot(kind='bar', stacked=True, figsize=(10,5))
plt.xlabel('Año del modelo')
plt.ylabel('Cantidad de autos')
plt.title('Distribución de autos por año de modelo y origen')
plt.legend(title='Origen', labels=['1: EE.UU.', '2: Europa', '3: Asia'])
plt.tight_layout()
plt.show()
```

### Conclusiones

- A mayor peso, el consumo (mpg) tiende a ser menor: hay una relación inversa.
- La distribución de `mpg` muestra que la mayoría de los autos consumen entre 15 y 30 mpg aproximadamente.
- El boxplot permite ver la dispersión de cilindros (4, 6 u 8) y el origen predominante.
- La mayoría de los autos en años tempranos son de EE.UU., y la proporción de autos europeos y asiáticos aumenta en los años más recientes del dataset.
